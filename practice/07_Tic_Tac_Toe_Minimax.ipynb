## Task 1. Create a 3 × 3 game board.

This block implements the task above. Variables from earlier blocks are reused.


```python
from math import inf
board=[" "]*9
lines=[(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
def show(b):
    for r in range(0,9,3): print(' | '.join(b[r:r+3]))

show(board)
```

## Task 2. Accept valid moves from the human player.

This block implements the task above. Variables from earlier blocks are reused.


```python
def human_move(b):
    while True:
        try:position=int(input("X: position 1-9: "))-1
        except ValueError:print("Enter a number.");continue
        if 0<=position<9 and b[position]==" ":
            b[position]="X"
            return
        print("Invalid position: choose an empty cell from 1 to 9.")
```

## Task 3. Implement Minimax to select the computer's move.

This block implements the task above. Variables from earlier blocks are reused.


```python
def minimax(b,maximizing):
    winner=result(b)
    if winner is not None: return {'O':1,'X':-1,'Draw':0}[winner]
    scores=[]
    for i in range(9):
        if b[i]==' ':
            b[i]='O' if maximizing else 'X'
            scores.append(minimax(b,not maximizing))
            b[i]=' '
    return max(scores) if maximizing else min(scores)

def computer_move(b):
    best_score,best_move=-inf,None
    for i in range(9):
        if b[i]==' ':
            b[i]='O'; score=minimax(b,False); b[i]=' '
            if score>best_score: best_score,best_move=score,i
    return best_move
```

## Task 4. Check for winning, losing and draw conditions.

This block implements the task above. Variables from earlier blocks are reused.


```python
def result(b):
    for i,j,k in lines:
        if b[i]!=' ' and b[i]==b[j]==b[k]: return b[i]
    return 'Draw' if ' ' not in b else None
```

## Task 5. Display the board after each move.

This block implements the task above. Variables from earlier blocks are reused.


```python
print("Human=X, Computer=O")
while result(board) is None:
    human_move(board)
    print("After human move:");show(board)
    if result(board) is None:
        board[computer_move(board)]="O"
        print("After computer move:");show(board)
print("Result:",result(board))
```

## Task 6. Demonstrate that the computer avoids moves that lead to a preventable loss.

This block implements the task above. Variables from earlier blocks are reused.


```python
threat=["X","X"," ","O"," "," "," "," "," "]
print("Defensive demonstration:");show(threat)
move=computer_move(threat)
assert move==2
threat[move]="O"
print("Computer blocks position",move+1);show(threat)
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. What is Minimax?**

A recursive game-tree algorithm that assumes both players choose their best available outcomes.

**4. What are maximizing and minimizing?**

O chooses the largest future score; X chooses the smallest score from O's perspective.

**5. What are terminal states?**

An X win, an O win, or a full board with no winner. They return -1, +1 or 0 respectively.

**6. Why undo each trial move?**

Otherwise a hypothetical move remains on the board and corrupts other branches of the search.

**7. How does the computer avoid a preventable loss?**

It chooses a move with the best worst-case score. In the demo it blocks position 3 rather than permitting X to win.

**8. Does optimal play always win?**

No. Tic-Tac-Toe is a draw with optimal play by both players. Minimax avoids a forced loss when a draw or win is available.

**9. How are invalid human moves handled?**

Non-numeric input, positions outside 1..9 and occupied positions are rejected.

**10. What is the branching factor?**

The number of legal moves from a state; initially 9, then decreasing.

**11. What is the complexity?**

General Minimax is O(b^d) time and O(d) recursion space with depth-first evaluation; this code also stores at most nine scores per call. Tic-Tac-Toe is small enough for full search.

**12. What is alpha-beta pruning?**

An optimization that skips branches unable to change the final minimax decision, while preserving the selected outcome.

**13. Does this version prefer the fastest win?**

Not explicitly: all wins score +1. A depth-adjusted score could prefer faster wins and slower losses.
