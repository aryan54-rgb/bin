## Task 1. Represent at least 8 cities as a weighted graph.

This block implements the task above. Variables from earlier blocks are reused.


```python
import heapq
cities = {"Pune":(0,0),"Mumbai":(0,3),"Nashik":(2,3),"Nagpur":(6,4),
          "Solapur":(3,0),"Kolhapur":(0,-3),"Satara":(0,-1),"Aurangabad":(4,3)}
graph = {"Pune":[("Mumbai",3),("Solapur",3),("Satara",1)],
         "Mumbai":[("Pune",3),("Nashik",2)],
         "Nashik":[("Mumbai",2),("Aurangabad",2)],
         "Nagpur":[("Aurangabad",3),("Solapur",7)],
         "Solapur":[("Pune",3),("Nagpur",7),("Kolhapur",6)],
         "Kolhapur":[("Satara",2),("Solapur",6)],
         "Satara":[("Pune",1),("Kolhapur",2)],
         "Aurangabad":[("Nashik",2),("Nagpur",3)]}
print("Weighted graph with",len(graph),"cities:")
for city,roads in graph.items():
    print(city,roads)
```

## Task 2. Assign edge costs representing distances and heuristic values representing estimated distances to the destination.

This block implements the task above. Variables from earlier blocks are reused.


```python
# Edge costs above are Manhattan distances between the practice coordinates.
def h(city, destination):
    return abs(cities[city][0]-cities[destination][0]) + abs(cities[city][1]-cities[destination][1])
print("Example heuristic estimates to Nagpur:")
for city in cities:
    print(city,h(city,"Nagpur"))
```

## Task 3. Implement A* using f(n)=g(n)+h(n)

This block implements the task above. Variables from earlier blocks are reused.


```python
def find_route(source,goal):
    queue=[(h(source,goal),0,source,[source])]
    best_cost={source:0}
    explored=0
    while queue:
        f,cost,city,path=heapq.heappop(queue)
        if cost!=best_cost[city]:
            continue
        explored+=1
        if city==goal:
            return path,cost,explored
        for neighbour,distance in graph[city]:
            new_cost=cost+distance
            if new_cost<best_cost.get(neighbour,float("inf")):
                best_cost[neighbour]=new_cost
                heapq.heappush(queue,(new_cost+h(neighbour,goal),new_cost,neighbour,path+[neighbour]))
    return [],float("inf"),explored
```

## Task 4. Accept the source and destination cities from the user.

This block implements the task above. Variables from earlier blocks are reused.


```python
print("Choose from:",", ".join(cities))
source=input("Source [Pune]: ").strip() or "Pune"
goal=input("Destination [Nagpur]: ").strip() or "Nagpur"
if source not in graph or goal not in graph:
    raise ValueError("Use a city from the printed list.")
```

## Task 5. Display the selected route, total path cost and nodes explored.

This block implements the task above. Variables from earlier blocks are reused.


```python
path,cost,explored=find_route(source,goal)
print("Route:"," -> ".join(path) if path else "No route")
print("Actual path cost:",cost)
print("Nodes explored:",explored)
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. What is the state?**

The current city. An edge is a road and its weight is the travel distance.

**4. What are g, h and f?**

g is actual cost from the source. h estimates remaining cost. A* uses f=g+h; Greedy uses h as its priority.

**5. What does heapq do?**

It implements a min-heap: heappop returns the queue entry with smallest priority.

**6. Why Manhattan distance here?**

Edge weights are Manhattan distances between artificial coordinates. The triangle inequality makes Manhattan-to-goal an admissible and consistent estimate for this graph.

**7. What is an admissible heuristic?**

One that never overestimates the true cheapest remaining cost. h(goal)=0. Our A* supports a better route being queued again.

**8. Is the route optimal?**

For this nonnegative graph and heuristic, A* returns an optimal route. Greedy has no general shortest-route guarantee even if h is admissible.

**9. What is the repeated-state check?**

A* skips queue entries whose cost is no longer best. Greedy skips already visited cities, preventing cycles.

**10. Why does changing the destination change h?**

h estimates distance to that specific goal, so it must be recomputed. Hard-coding estimates for one destination is incorrect for another.

**11. What if source equals destination?**

A single-city route with cost zero. This notebook counts the popped goal among explored nodes.

**12. What should the default example show?**

Pune to Nagpur has minimum cost 10. Routes and exploration counts may differ under tie-breaking. Greedy can happen to match A* without being generally optimal.

**13. What is the complexity?**

Search work depends on graph size and heuristic. For this finite graph with consistent h, heap operations give graph-search-style O((V+E) log(V+E)) work; carrying copied paths adds time and memory. General state-space search may grow exponentially.

**14. What is the easiest exam modification?**

Add cities/edges, select new endpoints and replace artificial distances with suitable costs. Use h=0 for A* if no safe heuristic is known; then it behaves like uniform-cost search.
