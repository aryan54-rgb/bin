## Task 1. Accept a Sudoku grid containing empty cells.

This block implements the task above. Variables from earlier blocks are reused.


```python
puzzle = [[5,3,0,0,7,0,0,0,0], [6,0,0,1,9,5,0,0,0], [0,9,8,0,0,0,0,6,0],
          [8,0,0,0,6,0,0,0,3], [4,0,0,8,0,3,0,0,1], [7,0,0,0,2,0,0,0,6],
          [0,6,0,0,0,0,2,8,0], [0,0,0,4,1,9,0,0,5], [0,0,0,0,8,0,0,7,9]]

# Optional manual entry: 9 rows, each containing 9 integers; 0 means empty.
# puzzle=[list(map(int,input("9 integers: ").split())) for _ in range(9)]
print("Initial Sudoku:")
for row in puzzle:print(*row)
```

## Task 2. Define the domain and row, column and 3 × 3 subgrid constraints.

This block implements the task above. Variables from earlier blocks are reused.


```python
def valid_sudoku(board, complete=False):
    if len(board)!=9 or any(len(row)!=9 for row in board): return False
    if any(type(x) is not int or not 0<=x<=9 for row in board for x in row): return False
    groups=board + [list(c) for c in zip(*board)]
    groups += [[board[r+i][c+j] for i in range(3) for j in range(3)]
               for r in (0,3,6) for c in (0,3,6)]
    return all(len([x for x in g if x])==len(set(g)-{0})
               and (not complete or set(g)==set(range(1,10))) for g in groups)

def safe(row,col,value):
    if value in puzzle[row]:return False
    if any(puzzle[r][col]==value for r in range(9)):return False
    for r in range(row//3*3,row//3*3+3):
        for c in range(col//3*3,col//3*3+3):
            if puzzle[r][c]==value:return False
    return True
valid_givens=valid_sudoku(puzzle)
print("Initial constraints valid:",valid_givens)
```

## Task 3. Implement a backtracking algorithm to assign valid values.

This block implements the task above. Variables from earlier blocks are reused.


```python
def solve():
    for row in range(9):
        for col in range(9):
            if puzzle[row][col]==0:
                for value in range(1,10):
                    if safe(row,col,value):
                        puzzle[row][col]=value
                        if solve():return True
                        puzzle[row][col]=0
                return False
    return True
```

## Task 4. Display the completed Sudoku grid.

This block implements the task above. Variables from earlier blocks are reused.


```python
found=valid_givens and solve()
if found:
    print("Completed Sudoku:")
    for row in puzzle:print(*row)
```

## Task 5. Validate the final solution against all constraints.

This block implements the task above. Variables from earlier blocks are reused.


```python
if found:
    assert valid_sudoku(puzzle,complete=True)
    print("Verified: every row, column and 3 x 3 box contains 1..9.")
```

## Task 6. Detect and report an unsolvable puzzle.

This block implements the task above. Variables from earlier blocks are reused.


```python
if not valid_givens:print("Invalid puzzle: the given digits violate constraints.")
elif not found:print("Unsolvable puzzle: no complete valid assignment exists.")
else:print("Solution found and validated.")
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. Variables and domains?**

Each empty cell is a variable with possible values 1 through 9.

**4. What are the constraints?**

Each row, column and 3x3 box must contain each digit 1..9 exactly once in a completed grid.

**5. How do we find a box?**

Its top-left cell is (row//3*3,column//3*3). Integer division groups row and column indices into blocks of three.

**6. Why check the initial puzzle?**

A board with duplicated givens is invalid even if all cells are filled. Never accept it merely because no empty cell exists.

**7. What is the base case?**

No unfilled cell remains. Because every assigned digit passed the checks, the constraints are satisfied; the program also validates the final grid.

**8. Why reset to zero?**

A later failure means the guessed digit must be removed before trying the next candidate.

**9. Can a valid-looking puzzle be unsolvable?**

Yes. Nonconflicting givens alone do not guarantee a completion. Exhaustive backtracking reports an unsolvable puzzle.

**10. Does it prove uniqueness?**

No. It returns one solution. To test uniqueness, count solutions and stop after finding two.

**11. Complexity?**

With E empty cells, a loose bound is O(9^E), with constant-size Sudoku checks. Recursion depth is at most E. Constraint propagation or choosing the smallest domain can reduce search.
