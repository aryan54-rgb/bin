## Task 1. Load a transactional dataset containing at least 10 transactions.

This block implements the task above. Variables from earlier blocks are reused.


````python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

transactions = [["milk","bread"],["bread","butter"],["milk","bread","butter"],
                ["milk","eggs"],["bread","eggs"],["milk","bread"],
                ["bread","butter","eggs"],["milk","bread","eggs"],
                ["bread","butter"],["milk","butter"]]
````

## Task 2. Convert the transactions into a suitable one-hot encoded representation.

This block implements the task above. Variables from earlier blocks are reused.


````python
encoder = TransactionEncoder()
values = encoder.fit(transactions).transform(transactions)
basket = pd.DataFrame(values,columns=encoder.columns_)
print("Transactions:",len(transactions))
print(basket)
````

## Task 3. Apply Apriori with a specified minimum support threshold.

This block implements the task above. Variables from earlier blocks are reused.


````python
frequent = apriori(basket,min_support=0.3,use_colnames=True)
print("Frequent itemsets:")
print(frequent.to_string(index=False))
````

## Task 4. Generate association rules using a specified minimum confidence threshold.

This block implements the task above. Variables from earlier blocks are reused.


````python
import inspect
settings = {"metric":"confidence","min_threshold":0.6}
if "num_itemsets" in inspect.signature(association_rules).parameters:
    settings["num_itemsets"] = len(basket)
rules = association_rules(frequent,**settings)
````

## Task 5. Calculate and display support, confidence and lift.

This block implements the task above. Variables from earlier blocks are reused.


````python
print("Association rules:")
print(rules[["antecedents","consequents","support","confidence","lift"]].to_string(index=False))

# Worked example from the 10 supplied baskets:
print("Milk support = 6/10 =",6/10)
print("Bread support = 8/10 =",8/10)
print("Milk and bread support = 4/10 =",4/10)
print("Milk -> bread confidence = 4/6 =",4/6)
print("Milk -> bread lift = (4/6)/(8/10) =",(4/6)/(8/10))
````

## Viva preparation guide

**1. What is an itemset?**

A set of products such as {milk,bread}. A transaction is one basket containing purchased products.

**2. What is support?**

support(A)=number of transactions containing A / total transactions. Rule support for A->B is support(A union B).

**3. What is confidence?**

confidence(A->B)=support(A union B)/support(A), estimating how often B occurs among baskets containing A.

**4. What is lift?**

lift(A->B)=confidence(A->B)/support(B). It compares co-occurrence with what would be expected under independence.

**5. Interpret lift values.**

Above 1: positive association; equal to 1: independence under this measure; below 1: negative association. Lift is not evidence of causation.

**6. Explain the milk -> bread example.**

Milk occurs 6 times, bread 8 times, both 4 times in 10 baskets. Support=0.4, confidence=4/6, lift=(4/6)/0.8≈0.833. High confidence alone does not imply lift above 1.

**7. What is Apriori's key property?**

Every subset of a frequent itemset must also be frequent. If an itemset is infrequent, all its supersets can be pruned.

**8. Why one-hot encode baskets?**

Each row is a transaction and each product column is True/False. Apriori needs presence of items rather than arbitrary category codes.

**9. What do the thresholds mean here?**

Minimum support 0.3 means at least 3 baskets out of 10. Minimum confidence 0.6 means at least 60% of antecedent baskets also contain the consequent.

**10. Are rules symmetric?**

Joint support is symmetric, but A->B confidence generally differs from B->A confidence because the denominators differ.

**11. How can I get no rules?**

Support or confidence thresholds may be too high, or there may be no frequent itemsets with two or more items. Inspect itemsets before lowering thresholds.

**12. What is the complexity?**

Candidate itemsets can grow exponentially with the number of distinct products. Apriori pruning helps but does not remove the worst-case explosion.

**13. Does this predict labels?**

No. It discovers associations in transactions. There is no supervised target y or classification accuracy.

**14. What should I print?**

Transactions/encoding, frequent itemsets and rules with antecedents, consequents, support, confidence and lift.
