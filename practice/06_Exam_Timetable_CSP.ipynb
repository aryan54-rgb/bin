## Create the CSP solver

A pair of examinations conflicts when at least one student takes both.


**Code layout:** 6 listed tasks → 6 code blocks. Run them in order. Imports and helper definitions are included in the relevant task. Where the PDF repeats a task number, these blocks follow its listed order. Learned preprocessing is fitted only after splitting.


## Task 1. Define at least 6 examinations, 4 time slots and a set of student enrollment records.

This block implements the task above. Variables from earlier blocks are reused.


```python
from itertools import combinations
exams=["DAA","AIML","DBMS","CN","OS","Math"]
slots=["T1","T2","T3","T4"]
enrollments={"S1":["DAA","AIML","DBMS"],"S2":["AIML","CN"],
             "S3":["DBMS","OS"],"S4":["OS","Math"],"S5":["DAA","Math"]}
print("Exams:",exams,"Slots:",slots)
```

## Task 2. Identify pairs of examinations that cannot occur simultaneously.

This block implements the task above. Variables from earlier blocks are reused.


```python
conflicts={exam:set() for exam in exams}
for student_exams in enrollments.values():
    for a,b in combinations(student_exams,2):
        conflicts[a].add(b)
        conflicts[b].add(a)
print("Conflicts:")
for exam in exams:print(exam,sorted(conflicts[exam]))
```

## Task 3. Define examination variables, available time slots and conflict constraints.

This block implements the task above. Variables from earlier blocks are reused.


```python
# Each exam is a variable; its domain is slots; neighbours must differ.
assigned={}
order=sorted(exams,key=lambda exam:-len(conflicts[exam]))
def safe(exam,slot):
    return all(assigned.get(other)!=slot for other in conflicts[exam])
print("Search order:",order)
```

## Task 4. Implement backtracking to generate a valid timetable.

This block implements the task above. Variables from earlier blocks are reused.


```python
def solve(index):
    if index==len(order):return True
    exam=order[index]
    for slot in slots:
        if safe(exam,slot):
            assigned[exam]=slot
            if solve(index+1):return True
            del assigned[exam]
    return False
found=solve(0)
```

## Task 5. Display the examination schedule and verify that no conflicts exist.

This block implements the task above. Variables from earlier blocks are reused.


```python
if found:
    for exam in exams:print(exam,"->",assigned[exam])
    assert all(assigned[a]!=assigned[b] for a in exams for b in conflicts[a])
    print("Verified: no conflicting examinations share a time slot.")
```

## Task 6. Report if the available time slots are insufficient for a valid schedule.

This block implements the task above. Variables from earlier blocks are reused.


```python
if not found:print("Insufficient time slots for a conflict-free schedule.")
else:print("A valid schedule was found with",len(set(assigned.values())),"used slots.")
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. What are variables, domains and constraints?**

Variables are exams; domains are the available slots; any two exams taken by a shared student must have different slots.

**4. Which graph problem does this model?**

Graph coloring. Exams are vertices, conflicts are edges, and slots are colors.

**5. How are conflict edges generated?**

For each student, create every pair of their enrolled exams using combinations(courses,2).

**6. Why use sets for conflicts?**

Repeated shared students create only one conflict edge per pair.

**7. Why sort exams by degree?**

Trying more constrained exams first often exposes impossible choices earlier. It is a search-order heuristic, not a correctness requirement.

**8. What is backtracking here?**

Try a slot absent from assigned conflicting neighbours; recurse; delete the assignment if it leads to failure.

**9. How is the result validated?**

For each conflict edge (a,b), check schedule[a]!=schedule[b].

**10. What happens with only two slots in this demo?**

DAA, AIML and DBMS form a triangle, so two slots are insufficient. Four available slots allow a valid schedule.

**11. Does a failed greedy attempt prove impossibility?**

No. This complete backtracking search can establish infeasibility for the given slots after exhausting all valid partial assignments.

**12. Complexity and limitations?**

Worst-case search is exponential in the number of exams, roughly S^E assignments for E exams and S slots, plus constraint checks. Room capacity and time preferences require additional constraints.
