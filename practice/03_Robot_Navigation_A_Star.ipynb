## Task 1. Create a grid of at least 8 × 8 cells.

This block implements the task above. Variables from earlier blocks are reused.


```python
import heapq
grid=[[0]*8 for _ in range(8)]
rows,cols=len(grid),len(grid[0])
print("Grid size:",rows,"x",cols)
```

## Task 2. Mark the starting cell, goal cell and obstacle cells.

This block implements the task above. Variables from earlier blocks are reused.


```python
start=(0,0)
goal=(7,7)
for row,col in [(1,1),(1,2),(1,3),(2,3),(3,3),(4,3),(5,3),(6,5)]:grid[row][col]=1
print("Start:",start,"Goal:",goal)
```

## Task 3. Implement A* Search using Manhattan distance as the heuristic.

This block implements the task above. Variables from earlier blocks are reused.


```python
def free(cell):
    r,c=cell
    return 0<=r<rows and 0<=c<cols and grid[r][c]==0
def h(cell):return abs(cell[0]-goal[0])+abs(cell[1]-goal[1])
def robot_path():
    if not free(start) or not free(goal):raise ValueError("Invalid start or goal")
    queue=[(h(start),0,start,[start])]
    best_cost={start:0}
    explored=0
    while queue:
        estimate,cost,cell,path=heapq.heappop(queue)
        if cost!=best_cost[cell]:continue
        explored+=1
        if cell==goal:return path,explored
        r,c=cell
        for next_cell in [(r+1,c),(r-1,c),(r,c+1),(r,c-1)]:
            new_cost=cost+1
            if free(next_cell) and new_cost<best_cost.get(next_cell,float("inf")):
                best_cost[next_cell]=new_cost
                heapq.heappush(queue,(new_cost+h(next_cell),new_cost,next_cell,path+[next_cell]))
    return [],explored
```

## Task 4. Display the shortest valid path on the grid.

This block implements the task above. Variables from earlier blocks are reused.


```python
path,explored=robot_path()
for r in range(rows):
    line=[]
    for c in range(cols):
        if (r,c)==start:symbol="S"
        elif (r,c)==goal:symbol="G"
        elif grid[r][c]:symbol="#"
        elif (r,c) in path:symbol="*"
        else:symbol="."
        line.append(symbol)
    print(" ".join(line))
print("Path:",path if path else "No path exists")
```

## Task 5. Calculate the path length and number of explored cells.

This block implements the task above. Variables from earlier blocks are reused.


```python
print("Path length in moves:",len(path)-1 if path else "No path")
print("Explored cells:",explored)
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. What is the state?**

A tuple (row,column), which can be used as a dictionary key.

**4. What are the actions?**

Four neighbouring moves: up, down, left, right. Out-of-grid and obstacle cells are rejected.

**5. Why Manhattan distance?**

abs(r-goal_r)+abs(c-goal_c) is a lower bound on the number of unit-cost four-direction moves. Obstacles can increase the real distance.

**6. How does A* choose a cell?**

Using the smallest g+h. g is moves already taken; h estimates remaining moves.

**7. Why a best_cost dictionary?**

It prevents unnecessary expansion of routes that are no cheaper than a known route to the same cell.

**8. Why len(path)-1?**

A path containing k cells has k-1 moves. A start already at the goal requires zero moves.

**9. What is the expected demo result?**

A valid path of 14 moves from (0,0) to (7,7). There may be several equally short paths.

**10. When does it fail?**

It reports no path after exhausting reachable states. A blocked or out-of-bounds endpoint is rejected before searching.

**11. What changes for diagonal moves?**

You must reconsider the heuristic and movement costs. Manhattan may overestimate if diagonal moves of cost 1 are permitted.

**12. Time and space?**

An R x C grid has at most RC states and about 4RC edges. Heap search is roughly O(RC log(RC)); explicitly copying paths in this teaching implementation adds overhead.
