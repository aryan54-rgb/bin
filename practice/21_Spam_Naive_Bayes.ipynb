## Task 1. Load a labeled spam message dataset.

This block implements the task above. Variables from earlier blocks are reused.


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
df=pd.read_csv("data/spam_practice.csv")
print("Synthetic practice dataset:",df.shape)
````

## Task 2. Clean the text and separate the message content from its class label.

This block implements the task above. Variables from earlier blocks are reused.


````python
X=df["message"].fillna("").str.strip().str.lower()
y=df["label"]
print("Class counts:",y.value_counts().to_dict())
````

## Task 3. Convert text into numerical features using CountVectorizer or TF-IDF.

This block implements the task above. Variables from earlier blocks are reused.


````python
# Define the vectorizer here; fit its vocabulary only after splitting.
vectorizer=TfidfVectorizer()
print("Text representation: TF-IDF word features")
````

## Task 4. Split the dataset into training and testing sets without fitting the vectorizer on test data.

This block implements the task above. Variables from earlier blocks are reused.


````python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
train_words=vectorizer.fit_transform(X_train)
test_words=vectorizer.transform(X_test)
print("Training vocabulary size:",len(vectorizer.vocabulary_))
````

## Task 5. Train a Multinomial Naïve Bayes classifier.

This block implements the task above. Variables from earlier blocks are reused.


````python
model=MultinomialNB()
model.fit(train_words,y_train)
````

## Task 6. Predict spam and non-spam labels for test messages.

This block implements the task above. Variables from earlier blocks are reused.


````python
prediction=model.predict(test_words)
messages=["Claim your free cash prize","Please send the class notes"]
print(pd.DataFrame({"Message":messages,"Prediction":model.predict(vectorizer.transform(messages))}))
````

## Task 7. Display precision, recall, F1-score and the confusion matrix.

This block implements the task above. Variables from earlier blocks are reused.


````python
print("Accuracy:",round(accuracy_score(y_test,prediction),4))
print("Class order:",model.classes_)
print("Confusion matrix:\n",confusion_matrix(y_test,prediction,labels=model.classes_))
print(classification_report(y_test,prediction,zero_division=0))
````

## Viva preparation guide



**1. What are X and y?**

X contains the predictor columns; y contains the target we want to predict. X is a 2D table and y is a 1D series.

**2. Why split training and testing data?**

Fit using the training examples. Evaluate on separate examples to estimate how well the model generalizes.

**3. What do fit() and predict() mean?**

fit() learns parameters from training examples. predict() applies the learned model to new feature rows.

**4. Why random_state=42?**

It makes random splits or model choices reproducible. The number 42 has no special mathematical meaning.

**5. What is data leakage?**

Information unavailable for a real prediction accidentally influences training. Examples: including the answer in X or fitting a scaler on test data. Fit learned preprocessing only on training data.

**6. What is overfitting?**

The model learns training-specific details/noise and does worse on unseen data. Control complexity and use validation or cross-validation.

**7. What is a hyperparameter?**

A setting chosen outside model fitting, such as max_depth, K or the number of trees. Choose it using training validation, not by repeatedly optimizing the final test result.

**8. How do I fix FileNotFoundError or KeyError?**

Open Jupyter in the praceice folder, keep data/ beside this notebook, and print df.columns.tolist() to check exact column names. Do not change X and y row counts separately.

**9. What does the confusion matrix contain?**

Rows are actual classes and columns are predicted classes, in the printed label order. For binary labels [0,1], it is [[TN,FP],[FN,TP]].

**10. Define accuracy, precision, recall and F1.**

Accuracy=(TP+TN)/N. Precision=TP/(TP+FP). Recall=TP/(TP+FN). F1=2PR/(P+R). classification_report prints per-class values and macro/weighted averages. zero_division=0 handles an undefined score.

**11. Why can accuracy be misleading?**

If most examples belong to one class, always predicting that class may give high accuracy. Inspect minority-class recall, precision, F1 and the confusion matrix.

**12. Why stratify=y?**

It approximately preserves class proportions in training and test splits. Each class needs enough examples.

**13. What is the Naive Bayes assumption?**

Word features are modeled as conditionally independent given the class. Real messages violate this simplification, but the model can still be useful.

**14. State Bayes theorem.**

P(class|features) is proportional to P(features|class)*P(class). Predict the class with greatest posterior score.

**15. What is TF-IDF?**

Term frequency multiplied by inverse document frequency. It downweights words common across many training documents; scikit-learn also applies its configured normalization.

**16. Why Multinomial NB?**

It works with nonnegative count-like features such as word counts and commonly with TF-IDF. Gaussian NB is intended for continuous features with Gaussian likelihoods.

**17. Why fit the vectorizer only on training messages?**

Vocabulary and IDF values are learned preprocessing. Test messages must not influence them.

**18. What is Laplace smoothing?**

Adding a positive smoothing value, alpha=1 by default here, avoids zero estimated probabilities for unseen word/class combinations.

**19. What does ham mean?**

A legitimate non-spam message.

**20. What limitation must be stated?**

This is a tiny synthetic practice dataset. Replace it with a sufficiently large real labeled dataset before making performance claims.
