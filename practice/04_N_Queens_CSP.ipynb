## Task 1. Accept the value of N from the user, with N at least 4.

This block implements the task above. Variables from earlier blocks are reused.


```python
n=int(input("Enter N (at least 4) [4]: ") or "4")
if n<4:raise ValueError("This question requires N >= 4")
```

## Task 2. Define the variables, domains and constraints.

This block implements the task above. Variables from earlier blocks are reused.


```python
# Variables: rows. Domain: columns 0..n-1. Constraints: no attacks.
columns=[-1]*n
def safe(row,col):
    for previous_row in range(row):
        if columns[previous_row]==col:return False
        if abs(columns[previous_row]-col)==row-previous_row:return False
    return True
print("Variables:",list(range(n)),"Domain:",list(range(n)))
```

## Task 3. Implement backtracking with constraint checking.

This block implements the task above. Variables from earlier blocks are reused.


```python
def solve(row):
    if row==n:return True
    for col in range(n):
        if safe(row,col):
            columns[row]=col
            if solve(row+1):return True
            columns[row]=-1
    return False
```

## Task 4. Display one valid board configuration.

This block implements the task above. Variables from earlier blocks are reused.


```python
found=solve(0)
if found:
    for col in columns:print(" ".join("Q" if c==col else "." for c in range(n)))
```

## Task 5. Verify that no two queens share the same row, column or diagonal.

This block implements the task above. Variables from earlier blocks are reused.


```python
if found:
    for row in range(n):
        for other in range(row):
            assert columns[row]!=columns[other]
            assert abs(columns[row]-columns[other])!=row-other
    print("Verified: no row, column or diagonal conflict.")
```

## Task 6. Report whether a solution was found.

This block implements the task above. Variables from earlier blocks are reused.


```python
print("Solution found:",found)
print("Queen columns:",columns if found else "No solution")
```

## Viva preparation guide

**1. How should I explain my program?**

State the input/state, allowed moves or constraints, the algorithm, stopping condition and printed result. Then trace one small example.

**2. How should I test it?**

Try the supplied example, a trivial solved case and an impossible or invalid case. Check the actual route/board/constraints, not just the final success message.

**3. What is a CSP?**

A constraint satisfaction problem has variables, domains and constraints. Find an assignment satisfying every constraint.

**4. Variables, domain and constraints here?**

Each row is a variable. Its domain is columns 0..N-1. Queens must have distinct columns and distinct diagonals. One queen per row is guaranteed by the representation.

**5. What is the diagonal test?**

Two queens attack diagonally when abs(column1-column2)==abs(row1-row2).

**6. What is backtracking?**

Choose a column, check constraints, recurse, and undo the choice if the remaining rows cannot be completed.

**7. What is the base case?**

row==N: all rows have a safe queen.

**8. What does col[r]=-1 do?**

It removes a failed assignment before trying another choice.

**9. What should N=4 show?**

One valid board, for example columns [1,3,0,2]. Another valid reflected configuration is equally correct.

**10. Does the program find all solutions?**

No. Returning True stops after the first valid arrangement. To enumerate all, collect each solution and continue searching.

**11. Why N>=4?**

The practical requires it. Mathematically N=1 is solvable and N=2 or N=3 is not.

**12. Complexity?**

Backtracking is exponential. With unique columns, the search has at most factorially many full permutations; this implementation also checks earlier rows for every candidate. Auxiliary assignment and recursion space are O(N).
